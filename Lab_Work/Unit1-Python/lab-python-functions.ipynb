In the previous exercise, you improved the code for managing customer orders by using loops and flow control. Now, let's take it a step further and refactor the code by introducing functions.

2. Define a function named `get_customer_orders` that takes no parameters. Inside the function, implement the code for prompting the user to enter the product names using a loop. The function should return the `customer_orders` set.

3. Define a function named `update_inventory` that takes `customer_orders` and `inventory` as parameters. Inside the function, implement the code for updating the inventory dictionary based on the customer orders.

4. Define a function named `calculate_order_statistics` that takes `customer_orders` and `products` as parameters.
```python
def initialize_inventory(products):
    """
    list of products
    using a loop and user input to create and return an inventory dictionary
    """
    inventory = {}

    for product in products:
        qty = int(input(f"Enter initial quantity for {product}: "))
        inventory[product] = qty

    return inventory 
```

```python
products = ["tshirt", "mug", "hat"]
inventory = initialize_inventory(products)

print(inventory)
```

```python
def get_customer_orders():
    """
    Takes no parameters.
    Using loop to prompt user for product names.
    Return a set of customer_orders.
    """
    customer_orders = set()

    while True:
        product= input("Enter name of product to order: ").lower()
        customer_orders.add(product)

        more = input("Add another product? (yes/no): ").lower()
        if more != "yes":
            break

    return customer_orders
```

```python
orders = get_customer_orders()

print(orders)
```

```python
def update_inventory(customer_orders, inventory):
    """
    Takes customer_orders (set) and inventory (dict).
    Inventory update based on customer orders
    Returns the updated inventory.
    """
    for product in customer_orders:
        if product in inventory:
            inventory[product] -= 1
        else:
            print(f"Out of stock:'{product}' is not in the inventory. ")

    return inventory
```

```python
inventory = {"t-shirt":5, "mug":3, "hat":4}
customer_orders = {"t-shirt", "mug"}

updated = update_inventory(customer_orders, inventory)

print(updated)
```

```python
def calculate_order_statistics(customer_orders, products):
    """
    Takes customer_orders (set) and products (list).
    total_products_ordered
    percentage_of_unique_products_ordered
    returns as dictionary: order_statistics.
    """
    total_products_ordered = len(customer_orders)

    valid_orders = [p for p in customer_orders if p in products]
    unique_products_ordered = len(valid_orders)

    total_unique_products = len(products)

    if total_unique_products == 0:
        percentage_of_unique_products_ordered = 0.0
    else:
        percentage_of_unique_products_ordered = (
            unique_products_ordered / total_unique_products
        ) * 100

    order_statistics = {
       "total_products_ordered": total_products_ordered,
    "percentage_of_unique_products_ordered": percentage_of_unique_products_ordered,
    }

    return order_statistics
```

```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
customer_orders = {"t-shirt", "mug", "unknown"}

stats = calculate_order_statistics(customer_orders, products)

print(stats)
```

```python
def print_order_statistics(order_statistics):
    """
    Takes order_statistics (dict)
    Prints the order statistics.
    """
    print("\n---Order Statistics---")
    print("Total products ordered:", order_statistics["total_products_ordered"])
    print(
        "Percentage of unique products ordered:",
        f"{order_statistics['percentage_of_unique_products_ordered']:.2f}%",
    )
    
```

```python
stats = {
    "total_products_ordered": 3,
    "percentage_of_unique_products_ordered": 40.0,
}

print_order_statistics(stats)
```

```python
def print_updated_inventory(inventory):
    """
    Takes inventory (dict).
    Prints the updated inventory
    """
    print("\n--- Updated Inventory ---")
    for product, qty in inventory.items():
        print(f"{product}: {qty}")
```

```python
inventory = {"t-shirt": 4, "mug": 2, "hat": 4, "book": 5, "keychain": 6}

print_updated_inventory(inventory)
```

```python
def main():

    products = ["t-shirt", "mug", "hat", "book", "keychain"]

    print("=== Initialize Inventory ===")
    inventory = initialize_inventory(products)

    print("\n=== Get Customer Orders ===")
    customer_orders = get_customer_orders()

    inventory = update_inventory(customer_orders, inventory)

    order_statistics = calculate_order_statistics(customer_orders, products)

    print_order_statistics(order_statistics)

    print_updated_inventory(inventory)


if __name__ == "__main__":
    main()
```
