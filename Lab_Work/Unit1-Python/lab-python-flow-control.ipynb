```python
## 1 . Look at your code from the lab data structures, and improve repeated code with loops

# Define a list of products

products = ["t-shirt", "mug", "hat", "book", "keychain"]
print("Available products:", products)

# Create an empty inventory dictionary
inventory = {}

# Ask the user to input the quantity of each product
print("\nEnter the quantity for each product:")

for item in products:
    while True:  
        quantity = input(f"Quantity for {item}: ")

        
        if quantity.isdigit():
            inventory[item] = int(quantity)
            break  
        else:
            print("Invalid input. Please enter a number.")

# Print final inventory
print("\nFinal inventory:")
print(inventory)
```

```python
#2. Prompt the user to enter the name of a product that a customer wants to order.

customer_orders = set()

print("\nAdd products to the customer's order.")
print("Type the product name exactly as in the product list.\n")

while True:
    # Ask user for a product name
    product = input("Enter a product to order: ").strip().lower()

    # Add product to the set
    customer_orders.add(product)
    print(f"Added '{product}' to the order.")

    # Ask if they want to add another product
    another = input("Do you want to add another product? (yes/no): ").strip().lower()

    # Stop if user says no
    if another != "yes":
        print("\nNo more products will be added.")
        break

# Print final customer order
print("\nCustomer orders:", customer_orders)
```

```python
#3. Instead of updating the inventory by subtracting 1 from the quantity of each product, 
#only do it for the products that were ordered (those in "customer_orders").

print("\nUpdating inventory based on customer orders...")

for item in customer_orders:
    # Check if the ordered product exists in the inventory
    if item in inventory:
        if inventory[item] > 0:
            inventory[item] -= 1
            print(f"Reduced inventory for {item}. New quantity: {inventory[item]}")
        else:
            print(f"Cannot reduce inventory for {item}. Stock is already 0.")
    else:
        print(f"'{item}' is not a valid product and cannot be updated.")
```
