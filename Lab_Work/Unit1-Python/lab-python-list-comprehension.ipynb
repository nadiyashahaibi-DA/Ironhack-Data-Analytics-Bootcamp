````python
def initialize_inventory(products):
    inventory = {product: int(input(f"Enter the quantity of {product}s available: ")) for product in products}
    
    return inventory
````

````python
def get_customer_orders():
    number_of_orders = int(input("Enter the number of customer orders: "))
    customer_orders = {
        input("Enter the product name to order: ").lower() 
        for i in range(number_of_orders)
    }
    
    return customer_orders
````

````python
def calculate_total_price(customer_orders):
    prices = {
        product: int(input(f"Enter the price for {product}: ")) 
        for product in customer_orders
    }
    
    total_price = sum(price for price in prices.values())
    return float(total_price)
````

````python
def update_inventory(inventory, customer_orders):
    new_inventory = {
        product: quantity - 1
        for product, quantity in inventory.items()
        if product not in customer_orders or quantity - 1 > 0
    }
    return new_inventory
````
