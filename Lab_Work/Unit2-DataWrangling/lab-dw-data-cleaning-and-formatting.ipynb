## Exercise 2: Cleaning invalid Values

The dataset contains columns with inconsistent and incorrect values that could affect the accuracy of our analysis. Therefore, we need to clean these columns to ensure that they only contain valid data.

Note that this exercise will focus only on cleaning inconsistent values and will not involve handling null values (NaN or None).

*Hint*:
- *Gender column contains various inconsistent values such as "F", "M", "Femal", "Male", "female", which need to be standardized, for example, to "M" and "F".*
- *State abbreviations be can replaced with its full name, for example "AZ": "Arizona", "Cali": "California", "WA": "Washington"*
- *In education, "Bachelors" could be replaced by "Bachelor"*
- *In Customer Lifetime Value, delete the `%` character*
- *In vehicle class, "Sports Car", "Luxury SUV" and "Luxury Car" could be replaced by "Luxury"*

```python
# Your code here
```

```python
# Checking Gender column

df["gender"].unique()
```

```python
# Gender column contains various inconsistent values such as "F", "M", "Femal", "Male", "female", which need to be standardized, for example, to "M" and "F".

# Clean inconsistent values
df["gender"] = df["gender"].replace({
    "Femal": "F",
    "female": "F",
    "Female": "F",
    "Male": "M",
    "male": "M"
})

# Convert to uppercase
df["gender"] = df["gender"].str.upper()

# Verify the result
df["gender"].unique()
```

```python
# Inspect current values
df["state"].unique()
```

```python
# Clean inconsistent values
df["state"] = df["state"].replace({
    "AZ": "Arizona",
    "Cali": "California",
    "WA": "Washington"
})

# Verify the result
df["state"].unique()
```

```python
# Inspect current values
df["education"].unique()
```

```python
# Replace Bachelors to Bachelor

df["education"] = df["education"].replace({
    "Bachelors": "Bachelor"
})

# Verify the result
df["education"].unique()
```

```python
# Checking In Customer Lifetime Value
df["customer_lifetime_value"].head()
```

```python
# Step 1: Remove the % symbol
df["customer_lifetime_value"] = df["customer_lifetime_value"].str.replace("%", "")

# Step 2: Convert to numeric
df["customer_lifetime_value"] = pd.to_numeric(df["customer_lifetime_value"])

# Step 3: Verify the result
df["customer_lifetime_value"].head()
df["customer_lifetime_value"].dtype
```

```python
# Inspect current values
df["vehicle_class"].unique()
```

```python
# replace by Luxury 

df["vehicle_class"] = df["vehicle_class"].replace({
    "Sports Car": "Luxury",
    "Luxury SUV": "Luxury",
    "Luxury Car": "Luxury"
})

# Verify the result
df["vehicle_class"].unique()
```

```python
# Your code here
```

```python
# Identify any columns with null or missing values.

# Count null values in each column
df.isnull().sum()

# Show only columns that contain nulls
df.isnull().sum()[df.isnull().sum() > 0]

```

```python
# 1: Drop rows with nulls
df_cleaned = df.dropna()
```

```python
# 2: Fill numerical nulls with mean
df["income"] = df["income"].fillna(df["income"].mean())

# 2: Fill categorical nulls with mode
df["gender"] = df["gender"].fillna(df["gender"].mode()[0])

# Verify
df.isnull().sum()
```

```python
# Fill the null values with the previous or next value in the column

# Forward fill
df["income"] = df["income"].ffill()

# Backward fill
df["gender"] = df["gender"].bfill()


df.isnull().sum()

```

```python
# After formatting data types, as a last step, convert all the numeric variables to integers.

# 1. Identify numeric columns
numeric_cols = df.select_dtypes(include=["float64", "int64", "object"]).columns

# 2. Convert everything to numeric (invalid values become NaN)
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors="coerce")

# 3. Fill NaN values
df[numeric_cols] = df[numeric_cols].fillna(0)

# 4. Convert to integer
df[numeric_cols] = df[numeric_cols].astype(int)

```

```python
# Verify 
df[numeric_cols].dtypes
```
